```python
from transformers import PreTrainedTokenizerFast
import numpy as np


def word_level_hidden_state(sentence_hidden_states: np.ndarray,
                            sentence: str,
                            word_position: int,
                            tokenizer: PreTrainedTokenizerFast,
                            strategy: str = "avg") -> np.ndarray:

    tokenizer_output = tokenize(
        tokenizer=tokenizer,
        sentences=[sentence],
    )

    words = sentence.split(" ")
    word = words[word_position]
    word_index = sentence.index(word)

    offset_mappings = tokenizer_output["offset_mapping"][0].tolist()
    offset_mapping_indices = []

    for i, offset_mappings in enumerate(offset_mappings):
        if offset_mappings[0] == offset_mappings[1] and offset_mappings[0] == 0:
            if i == 0:
                print("Skipp heading service token")
            else:
                print("Skipp trailing service token")
            continue
        if offset_mappings[0] >= word_index - 1:
            if word_position == len(words) - 1:
                offset_mapping_indices.append(i)
            else:
                next_word = words[word_position + 1]
                next_word_index = sentence.index(next_word)
                if offset_mappings[1] <= next_word_index:
                    offset_mapping_indices.append(i)
                else:
                    break

    if strategy == "avg":
        return sentence_hidden_states[offset_mapping_indices, :, :].mean(axis=0)
    else:
        raise ValueError("Strategy not implemented")
```

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch
import math

# model_name = "gpt2"  # можно заменить на любую causal LM
# tokenizer = AutoTokenizer.from_pretrained(model_name)
# model = AutoModelForCausalLM.from_pretrained(model_name)
model.eval()

def compute_surprisal(text):
    # Токенизация
    inputs = tokenizer(text, return_tensors="pt")
    input_ids = inputs["input_ids"]

    # Прогон модели
    with torch.no_grad():
        outputs = model(input_ids, labels=input_ids)
        # logits: [batch, seq_len, vocab_size]
        logits = outputs.logits

    # Сдвигаем токены, чтобы каждый логит предсказывал следующий токен
    shifted_logits = logits[:, :-1, :]
    shifted_labels = input_ids[:, 1:]

    # Вычисляем распределение вероятностей
    log_probs = torch.log_softmax(shifted_logits, dim=-1)

    # Берём log p для правильных токенов
    token_log_probs = log_probs.gather(2, shifted_labels.unsqueeze(-1)).squeeze(-1)

    # surprisal = -log p
    surprisals = -token_log_probs

    # Декодируем токены для удобства
    tokens = tokenizer.convert_ids_to_tokens(shifted_labels[0])

    return list(zip(tokens, surprisals[0].tolist()))

# Пример
texts = ["Автобусы проходят массовую дезинфекцию", "Автобусы проходят массовую дезинфекцией"]

results = []

for text in texts:

    result = compute_surprisal(text)
    results.append(result)

    for tok, s in result:
        print(f"{tok:15} surprisal = {s:.4f}")
```
