```python
import mne
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.patches as patches
import os

# ======================
# 0. Настройки
# ======================
N_SUBJECTS = 21  # p0 ... p20
FILES = [f"../hf_datasets/eeg/p{i}_epochs.fif" for i in range(N_SUBJECTS)]
BASELINE = (-0.4, 0)

# Категории стимулов
congruent_ids = ['Stimulus/S  1_1','Stimulus/S  1_2','Stimulus/S  1_3']
sem_ids       = ['Stimulus/S  2_1','Stimulus/S  2_2','Stimulus/S  2_3']
gram_ids      = ['Stimulus/S  3_1','Stimulus/S  3_2','Stimulus/S  3_3']
both_ids      = ['Stimulus/S  4_1','Stimulus/S  4_2','Stimulus/S  4_3']

# Списки ERP всех участников
evk_cong_all = []
evk_sem_all  = []
evk_gram_all = []
evk_both_all = []

# ======================
# 1. Обработка каждого файла и ERP по человеку
# ======================
for file in FILES:
    if not os.path.exists(file):
        print(f"Файл не найден: {file}")
        continue

    print(f"Загружаю {file}")
    epochs = mne.read_epochs(file, preload=True)
    epochs.apply_baseline(BASELINE)

    # Разделяем условия
    e_cong = epochs[congruent_ids]
    e_sem  = epochs[sem_ids]
    e_gram = epochs[gram_ids]
    e_both = epochs[both_ids]

    # Усредняем
    evk_cong_all.append(e_cong.average().pick_channels(['Fz', 'FCz', 'Pz', 'Oz']))
    evk_sem_all.append(e_sem.average().pick_channels(['Fz', 'FCz', 'Pz', 'Oz']))
    evk_gram_all.append(e_gram.average().pick_channels(['Fz', 'FCz', 'Pz', 'Oz']))
    evk_both_all.append(e_both.average().pick_channels(['Fz', 'FCz', 'Pz', 'Oz']))

print("\n=== Все участники обработаны ===")

# ======================
# 2. Гранд-Авередж по условиям
# ======================
evk_cong_GA = mne.grand_average(evk_cong_all)
evk_sem_GA  = mne.grand_average(evk_sem_all)
evk_gram_GA = mne.grand_average(evk_gram_all)
evk_both_GA = mne.grand_average(evk_both_all)

print("\n=== Созданы Grand Average ERP ===")

# ======================
# 3. Разностные волны
# ======================
evk_sem_minus_cong  = mne.combine_evoked([evk_sem_GA,  evk_cong_GA],  [1, -1])
evk_gram_minus_cong = mne.combine_evoked([evk_gram_GA, evk_cong_GA], [1, -1])
evk_both_minus_cong = mne.combine_evoked([evk_both_GA, evk_cong_GA], [1, -1])

gg = mne.combine_evoked([evk_sem_GA, evk_gram_GA, evk_cong_GA], [1, 1, -2])
gg1 = mne.combine_evoked([evk_sem_GA, evk_gram_GA, evk_both_GA, evk_cong_GA], [1, 1, -1, -1])

# ======================
# 4. Функция: heatmap с рамкой
# ======================
def plot_with_box(evoked, title, t_start, t_end, color='red'):
    data = evoked.data
    times = evoked.times
    n_chan = data.shape[0]

    fig, ax = plt.subplots(figsize=(10, 6))
    im = ax.imshow(
        data,
        aspect='auto',
        origin='lower',
        extent=[times[0], times[-1], 0, n_chan]
    )
    plt.colorbar(im, ax=ax)
    ax.set_title(title)
    ax.set_xlabel("Time (s)")
    ax.set_ylabel("Channels")

    rect = patches.Rectangle(
        (t_start, 0),
        t_end - t_start,
        n_chan,
        linewidth=2,
        edgecolor=color,
        facecolor='none'
    )
    ax.add_patch(rect)

    plt.tight_layout()
    plt.show()


# ======================
# 5. Heatmaps (каналы × время)
# ======================
plot_with_box(evk_sem_GA,  "GA Semantic — N400", 0.30, 0.50)
plot_with_box(evk_gram_GA, "GA Grammar — P600", 0.50, 0.70)
plot_with_box(evk_both_GA, "GA Both — N400", 0.30, 0.50)
plot_with_box(evk_both_GA, "GA Both — P600", 0.50, 0.70)
plot_with_box(gg, "gg", 0.50, 0.70)
plot_with_box(gg1, "gg1", 0.50, 0.70)

# ======================
# 6. Topomap
# ======================
times_of_interest = [0.4, 0.6]

evk_cong_GA.plot_topomap(times=times_of_interest, ch_type='eeg', time_unit='s',
    title="GA Topomap — Congruent (0.4s, 0.6s)")
evk_sem_GA.plot_topomap(times=times_of_interest, ch_type='eeg', time_unit='s',
    title="GA Topomap — Semantic (N400 & P600)")
evk_gram_GA.plot_topomap(times=times_of_interest, ch_type='eeg', time_unit='s',
    title="GA Topomap — Grammar (N400 & P600)")
evk_both_GA.plot_topomap(times=times_of_interest, ch_type='eeg', time_unit='s',
    title="GA Topomap — Both (N400 & P600)")

# # ======================
# # 7. Topomap разностей
# # ======================
# evk_sem_minus_cong.plot_topomap(times=times_of_interest, ch_type='eeg', time_unit='s',
#     title="GA Diff Topomap — Semantic − Congruent")
# evk_gram_minus_cong.plot_topomap(times=times_of_interest, ch_type='eeg', time_unit='s',
#     title="GA Diff Topomap — Grammar − Congruent")
# evk_both_minus_cong.plot_topomap(times=times_of_interest, ch_type='eeg', time_unit='s',
#     title="GA Diff Topomap — Both − Congruent")
#
# print("\n=== Готово! Все графики построены ===")
```

```python
# ============================================
# 8. Difference heatmaps (каналы × время)
# ============================================

def plot_diff_heatmap(evoked_diff, title, highlight=None):
    """
    evoked_diff: MNE Evoked object (difference wave)
    highlight: (t_start, t_end) or None
    """

    data = evoked_diff.data
    times = evoked_diff.times
    n_chan = data.shape[0]

    fig, ax = plt.subplots(figsize=(10, 6))
    im = ax.imshow(
        data,
        aspect='auto',
        origin='lower',
        extent=[times[0], times[-1], 0, n_chan]
    )
    plt.colorbar(im, ax=ax)
    ax.set_title(title)
    ax.set_xlabel("Time (s)")
    ax.set_ylabel("Channels")

    # Highlight a time window
    if highlight is not None:
        t_start, t_end = highlight
        rect = patches.Rectangle(
            (t_start, 0),
            t_end - t_start,
            n_chan,
            linewidth=2,
            edgecolor='red',
            facecolor='none'
        )
        ax.add_patch(rect)

    plt.tight_layout()
    plt.show()


# ---- Semantic – Congruent (N400 reason)
plot_diff_heatmap(
    evk_sem_minus_cong.pick_channels(['Fz', 'FCz', 'Pz', 'Oz']),
    "GA Difference Heatmap — Semantic − Congruent",
    highlight=(0.30, 0.50)
)

# ---- Grammar – Congruent (P600 reason)
plot_diff_heatmap(
    evk_gram_minus_cong.pick_channels(['Fz', 'FCz', 'Pz', 'Oz']),
    "GA Difference Heatmap — Grammar − Congruent",
    highlight=(0.50, 0.70)
)

# ---- Both – Congruent (N400 + P600)
plot_diff_heatmap(
    evk_both_minus_cong.pick_channels(['Fz', 'FCz', 'Pz', 'Oz']),
    "GA Difference Heatmap — Both − Congruent (N400 window)",
    highlight=(0.30, 0.50)
)

plot_diff_heatmap(
    evk_both_minus_cong,
    "GA Difference Heatmap — Both − Congruent (P600 window)",
    highlight=(0.50, 0.70)
)
```

```python
import numpy as np

def find_roi(evoked_diff, tmin, tmax, top_k=8):
    """
    Ищет ROI каналы в указанных временных окнах difference wave.
    top_k — сколько самых «сильных» каналов включить.
    """

    times = evoked_diff.times
    idx = np.where((times >= tmin) & (times <= tmax))[0]

    # Усреднение за окно времени (чтобы убрать шум)
    data = evoked_diff.data[:, idx].mean(axis=1)

    # Находим каналы с max абсолютным эффектом
    strongest_idx = np.argsort(np.abs(data))[-top_k:]

    roi_channels = [evoked_diff.ch_names[i] for i in strongest_idx]

    return roi_channels, data[strongest_idx]
```

```python
roi_sem_N400, vals = find_roi(evk_sem_minus_cong, 0.30, 0.50)
roi_gram_P600, vals2 = find_roi(evk_gram_minus_cong, 0.50, 0.70)

print("Semantic N400 ROI channels:", roi_sem_N400)
print("Grammar P600 ROI channels:", roi_gram_P600)
```

```python
def find_roi_topomap(evoked_diff, tmin, tmax, percentile=75):
    times = evoked_diff.times
    idx = np.where((times >= tmin) & (times <= tmax))[0]

    mean_amp = evoked_diff.data[:, idx].mean(axis=1)
    thresh = np.percentile(np.abs(mean_amp), percentile)

    roi = [evoked_diff.ch_names[i]
           for i,v in enumerate(mean_amp)
           if abs(v) >= thresh]

    return roi
```

```python
roi_sem_N400 = find_roi_topomap(evk_sem_minus_cong, 0.30, 0.50).pick_channels(['Fz', 'FCz', 'Pz', 'Oz'])
roi_gram_P600 = find_roi_topomap(evk_gram_minus_cong, 0.50, 0.70).pick_channels(['Fz', 'FCz', 'Pz', 'Oz'])

print(roi_sem_N400)
print(roi_gram_P600)
```

```python
from mne.stats import permutation_cluster_test

data_sem  = np.array([e.data for e in evk_sem_all.pick_channels(['Fz', 'FCz', 'Pz', 'Oz'])])   # subjects × ch × time
data_cong = np.array([e.data for e in evk_cong_all.pick_channels(['Fz', 'FCz', 'Pz', 'Oz'])])

X = [data_sem, data_cong]

T_obs, clusters, p_values, H0 = permutation_cluster_test(
    X, n_permutations=2000, tail=0, n_jobs=1)
```

```python
def plot_roi_erp(evoked, roi, title):
    picks = [evoked.ch_names.index(ch) for ch in roi]
    data = evoked.data[picks].mean(axis=0)  # среднее по каналам ROI

    plt.figure(figsize=(8,4))
    plt.plot(evoked.times, data, linewidth=2)
    plt.axvline(0, color='k', linestyle='--')
    plt.title(title)
    plt.xlabel("Time (s)")
    plt.ylabel("Amplitude (µV)")
    plt.grid(True)
    plt.show()

# пример:
plot_roi_erp(evk_sem_minus_cong.pick_channels(['Fz', 'FCz', 'Pz', 'Oz']), roi_sem_N400.pick_channels(['Fz', 'FCz', 'Pz', 'Oz']),
             "ERP (ROI) — Semantic − Congruent (N400 ROI)")
```
