```python
epochs_dict: dict[str, mne.Epochs] = {}

for file_name in os.listdir("../hf_datasets/eeg"):
    if file_name.endswith(".fif"):
        epochs = mne.read_epochs(os.path.join("../hf_datasets/eeg", file_name))
        epochs = epochs.apply_baseline((-0.4, 0.0))
        epochs._data *= -1
        epochs_dict[file_name.split("_")[0]] = epochs
```

```python
threshold = 550
filtered_epoch_dict: dict[str, mne.Epochs] = {}
for k, epochs in epochs_dict.items():
    if len(epochs) > threshold:
        filtered_epoch_dict[k] = epochs

len(filtered_epoch_dict)
```

Далее необходимо сопоставить события для субъектов и оставить те, которые присутствуют в каждой эпохе

С датасетом **ContributorsSIGNAL/SIGNAL** есть проблема
1) В файлах эпох лежат данные для испытуемых с отфильтрованными стимулами (<= 600 записей)
2) В файлах информациях об эпохах лежат данные обо всех стимулах (600) и их временные метки
3) Временные метки из .fif файлов не совпадают с метками из .csv

Я взял датасет из черновика статьи **zhuravlevahana/SIGNAL**
1) для первых двух пациентов epoch_events совпали с epoch_info по колонке start
2) для остальных по колонке time2
3) это позволило получить связь с тем какие предложения остались в ЭЭГ после фильтрации, для дальнейшего анализа (например построения RSA)

Видимо, формат файлов epoch_info для чернового варианта набора данных менялся, а вот .fif файлы
похоже переносились без изменений, поэтому сейчас итоговоый набор данных не совсем валиден

```python
stimuli_df = pd.read_csv("../hf_datasets/stimuli.csv")

dataset_folder = "../hf_datasets/"

def align_events(dataset_folder: str, aligned_files_output: str) -> None:

    os.makedirs(aligned_files_output, exist_ok=True)

    xlsx_files = []
    fif_files = []
    for file_name in os.listdir(dataset_folder):
        if str(file_name).endswith(".xlsx"):
            xlsx_files.append(str(os.path.join(dataset_folder, file_name)))
        if str(file_name).endswith(".fif"):
            fif_files.append(str(os.path.join(dataset_folder, file_name)))

    fif_p_dict: dict[str, str] = {}
    for file_name in fif_files:
        fif_p_name = file_name.split("/")[-1].split("_")[0]
        fif_p_dict[fif_p_name] = file_name

    for xlsx_file in xlsx_files:
        xlsx_p_name = xlsx_file.split("/")[-1].split("_")[0]
        fif_file = fif_p_dict[xlsx_p_name]

        print(xlsx_file, fif_file)

        if xlsx_p_name in {"p0", "p1"}:
            col = "start"
        else:
            col = "time2"

        xlsx_df = pd.read_excel(xlsx_file)
        from ast import literal_eval
        xlsx_df["sentence_combined"] = xlsx_df["sentence"].apply(lambda x: " ".join(literal_eval(x)))

        epoch = mne.read_epochs(fif_file, preload=True)
        event_id_dict = dict(epoch.event_id)
        epochs_events = pd.DataFrame(epoch.events, columns=[col, "1", "event_id"])

        try:
            merged_df = epochs_events.merge(xlsx_df, how="left", on=col)
        except Exception as e:
            print(e)
            print(xlsx_file)
            break

        for i, row in merged_df.iterrows():
            stimuli_id = row["event_id"]
            stimuli_name = row["name"]
            if event_id_dict[stimuli_name] != int(stimuli_id):
                print(row)
                raise RuntimeError("Stimuli are not equal!")

        merged_df = merged_df.rename(columns={col: "event_time"})
        merged_df.to_csv(aligned_files_output + "/" + f"{xlsx_p_name}_aligned.csv", index=False)

aligned_event_files_output = "../eeg_epochs_aligned"
align_events(dataset_folder="../hf_old_datasets", aligned_files_output=aligned_event_files_output)
```

Убедимся, что количество строк в выровненных файлах совпадает с количеством событий в отфильтрованных участниках. Также определим количество стимулов, присутствующих во всех замерах для субъектов.

```python
import pathlib

# Дадим каждому предложению уникальный ID, составим
stimuli_df["absolute_sentence_id"] = stimuli_df.index.copy().tolist()
common_stimuli_set = set(stimuli_df["absolute_sentence_id"])

for k, epoch in filtered_epoch_dict.items():
    aligned_file_path = pathlib.Path(aligned_event_files_output) / f"{k}_aligned.csv"
    epochs_aligned_events = pd.read_csv(aligned_file_path)
    if len(epochs_aligned_events) != len(epoch):
        raise RuntimeError("Number of events in epoch file does not match!")

    merged = epochs_aligned_events.merge(stimuli_df, how="left", left_on="sentence_combined", right_on="sentence")

    common_stimuli_set = common_stimuli_set.intersection(set(merged["absolute_sentence_id"]))

print(f"All rows are matched! Common stimuli set size: {len(common_stimuli_set)}")
```

Добавим типы событий в набор данных стимулов

```python
events_description_df = pd.read_csv("../hf_datasets/events_description.csv")

def map_congruence_conditions(condition: str):
    condition = condition.lower()
    if condition == "congruent":
        return "normal"
    elif condition == "semantically incongruent":
        return "semantics"
    elif condition == "grammatically incongruent":
        return "grammar"
    elif condition == "semantically-grammatically incongruent":
        return "semantics_grammar"
    else:
        raise ValueError("Unknown condition")

events_description_df["target"] = events_description_df["congruence_condition"].apply(map_congruence_conditions)
events_description_df["sentence_structure"] = events_description_df["sentence_structure"].apply(lambda x: x.replace("Adjective", "Adj").replace("Genitive", "Gen"))

events_description_df = events_description_df[["name", "sentence_structure", "target"]]
events_description_df.rename(columns={"name": "event_name", "sentence_structure": "structure"}, inplace=True)

events_description_df
```

```python
common_stimuli_df = common_stimuli_df.merge(
    events_description_df,
    how="left",
    on=["structure", "target"],
)
```

Предварительно привяжем id предложений к каждому событию в epoch файле

```python

for epoch_name, epoch in tqdm(filtered_epoch_dict.items(), desc="Fill sentence id metadata"):

    aligned_eeg_events_df = pd.read_csv(aligned_event_files_output + "/" + f"{epoch_name}_aligned.csv").reset_index(drop=True)

    event_time_and_sentence_id = aligned_eeg_events_df.merge(
        stimuli_df[["sentence", "absolute_sentence_id"]],
        how="left",
        left_on=["sentence_combined"],
        right_on=["sentence"],
    )[["event_time", "absolute_sentence_id"]].reset_index(drop=True)

    epoch_events_df = pd.DataFrame(epoch.events).reset_index(drop=True)

    aligned_metadata = epoch_events_df.merge(
        event_time_and_sentence_id,
        how="left",
        left_on=0,
        right_on="event_time",
    )[["absolute_sentence_id"]].rename(columns={"absolute_sentence_id": "sent_id"}).reset_index(drop=False)

    epoch.metadata = aligned_metadata
```

Для каждого субъекта достанем записи ЭЭГ согласно `common_stimuli_set`, приведем к одному порядку и посчитаем Time-Resolved RDM.

```python
for epochs_name, epochs in filtered_epoch_dict.items():
    metadata: pd.DataFrame = epochs.metadata
    metadata_surrogate_index = list(range(0, len(metadata)))
    indices = [
        metadata_surrogate_index[x] for x in common_stimuli_set if x in metadata_surrogate_index]

    epochs_sel = epochs[indices]
    epochs_sel = epochs[indices].copy()

    print(epochs_sel)

    rdms = mne_rsa.rdm_epochs(
        epochs_sel,
        dist_metric="correlation",
        temporal_radius=0.25,
        picks=["Fz", "FCz", "Pz"],
        tmin=0.0,
        tmax=0.8
    )

    print(rdms)

    w = 100
    fig = mne_rsa.plot_rdms(
        list(rdms)[0],
    )
    fig.set_size_inches(3, 3)

    break


    break
```
