```python
epochs_list = []

for file_name in os.listdir("../hf_datasets/eeg"):
    if file_name.endswith(".fif"):
        epochs = mne.read_epochs(os.path.join("../hf_datasets/eeg", file_name))
        epochs = epochs.apply_baseline((-0.4, 0.0))
        epochs_list.append(epochs)
```

```python
s_1_x_stimuli = [
    "Stimulus/S  1_1",
    "Stimulus/S  1_2",
    "Stimulus/S  1_3",
]

s_2_x_stimuli = [
    "Stimulus/S  2_1",
    "Stimulus/S  2_2",
    "Stimulus/S  2_3",
]

s_3_x_stimuli = [
    "Stimulus/S  3_1",
    "Stimulus/S  3_2",
    "Stimulus/S  3_3",
]

s_4_x_stimuli = [
    "Stimulus/S  4_1",
    "Stimulus/S  4_2",
    "Stimulus/S  4_3",
]

s_1_x_erps = [s[s_1_x_stimuli].average() for s in epochs_list]
s_2_x_erps = [s[s_2_x_stimuli].average() for s in epochs_list]
s_3_x_erps = [s[s_3_x_stimuli].average() for s in epochs_list]
s_4_x_erps = [s[s_4_x_stimuli].average() for s in epochs_list]

s_1_x_erps_ga: mne.EvokedArray = mne.grand_average(s_1_x_erps)
s_2_x_erps_ga: mne.EvokedArray = mne.grand_average(s_2_x_erps)
s_3_x_erps_ga: mne.EvokedArray = mne.grand_average(s_3_x_erps)
s_4_x_erps_ga: mne.EvokedArray = mne.grand_average(s_4_x_erps)
```

```python
evokeds = [s_1_x_erps_ga, s_2_x_erps_ga, s_3_x_erps_ga, s_4_x_erps_ga]
labels  = ["Cong", "Semantic", "Grammatic", "Both"]

# channels = ['Fz', 'FCz', 'Pz', 'Oz']
channels = ['Fz']

fig, ax = plt.subplots(figsize=(10, 6))

ax.hlines(xmin=300, xmax=500, y=-7e-7, color="k", linestyle="dashed")

for evk, label in zip(evokeds, labels):
    data = evk.copy().pick(channels).data.mean(axis=0)  # среднее по выбранным каналам
    ax.plot(evk.times * 1000, data, label=label)


ax.set_title("Grand averages for selected channels")
ax.set_xlabel("Time (ms)")
ax.set_ylabel("Amplitude (µV)")
ax.legend()
plt.grid(True)
plt.show()
```

```python
from collections import defaultdict

evokeds_by_event = defaultdict(list)

for epochs in epochs_list:
    evokeds = epochs.average(by_event_type=True)
    for evoked in evokeds:
        if evoked.comment not in evokeds_by_event:
            evokeds_by_event[evoked.comment] = []
        evokeds_by_event[evoked.comment].append(evoked)
```

```python
conditions = list(evokeds_by_event.keys())

cmap = plt.get_cmap('tab20')
colors = {cond: cmap(i) for i, cond in enumerate(conditions)}

mne.viz.plot_compare_evokeds(
    evokeds_by_event,  # dict: {'event1': [evoked_subj1, evoked_subj2, ...]}
    picks=['Fz', "FCz", "Pz"],
    combine='mean',
    ci=0.95,
    colors=colors,
)
```
