```python
epochs_list = []

for file_name in os.listdir("../hf_datasets/eeg"):
    if file_name.endswith(".fif"):
        epochs = mne.read_epochs(os.path.join("../hf_datasets/eeg", file_name))
        epochs = epochs.apply_baseline((-0.4, 0.0))
        epochs._data *= -1
        epochs_list.append(epochs)
```

```python
conditions = list(epochs_list[0].event_id.keys())

evokeds_list = []

for i, epoch in enumerate(epochs_list):
    evokeds = {}
    for cond in conditions:
        epochs_cond = epoch[cond]
        if len(epochs_cond) == 0:
            continue
        evokeds[cond] = epochs_cond.average()
    evokeds_list.append(evokeds)
```

```python
events_description_df = pd.read_csv("/Users/ilyamikheev/Desktop/cuscience/SIGNAL-analysis/hf_datasets/events_description.csv")

def map_congruence_conditions(condition: str):
    condition = condition.lower()
    if condition == "congruent":
        return "normal"
    elif condition == "semantically incongruent":
        return "semantics"
    elif condition == "grammatically incongruent":
        return "grammar"
    elif condition == "semantically-grammatically incongruent":
        return "semantics_grammar"
    else:
        raise ValueError("Unknown condition")

events_description_df["target"] = events_description_df["congruence_condition"].apply(map_congruence_conditions)
events_description_df["sentence_structure"] = events_description_df["sentence_structure"].apply(lambda x: x.replace("Adjective", "Adj").replace("Genitive", "Gen"))

events_description_df = events_description_df[["name", "sentence_structure", "target"]]

events_description_df
```

```python
aligned_conditions = pd.DataFrame({
    "name": conditions,
})

aligned_conditions = aligned_conditions.merge(events_description_df, how="left", on="name")

aligned_conditions
```

```python
model_hidden_state_folders = [
    ("ai-sage_GigaChat3-10B-A1.8B-base", "/Users/ilyamikheev/Desktop/cuscience/SIGNAL-analysis/src/results/ai-sage_GigaChat3-10B-A1.8B-base")
]

model_rdm_dict: dict[str, list] = {}

stimuli_df = pd.read_csv("/Users/ilyamikheev/Desktop/cuscience/SIGNAL-analysis/hf_datasets/stimuli.csv")
for model, folder in model_hidden_state_folders:
    results_df = pd.read_csv(folder + "/results.csv")
    results_df = results_df.merge(stimuli_df, how="left", left_on="sentence", right_on="sentence")

    agg_groups = results_df[["structure", "target"]].groupby(["structure", "target"]).groups

    hiddens = []
    for i, row in results_df.iterrows():
        relative_hidden_state_pathstring = row["hidden_states_path"].replace(".npy", "_word_level_hidden_states.npy")
        relative_hidden_state_path = pathlib.Path(relative_hidden_state_pathstring)
        absolute_hidden_state_path = pathlib.Path(folder).parent.parent / relative_hidden_state_path
        hiddens.append(np.load(absolute_hidden_state_path))
    hiddens = np.stack(hiddens, axis=0)

    group_mean_hidden_state_dict = {}
    for group, indices in agg_groups.items():
        hidden_states_to_agg = hiddens[indices, :, :]
        group_mean_hidden_state_dict[group] = hidden_states_to_agg.mean(axis=0)

    ordered_hidden_states = []
    for _, aligned_condition_row in aligned_conditions.iterrows():
        event_type = aligned_condition_row["name"]
        group_key = (aligned_condition_row["sentence_structure"], aligned_condition_row["target"])
        for key, values in group_mean_hidden_state_dict.items():
            structure = key[0]
            target = key[1]
            if structure == aligned_condition_row["sentence_structure"] and target == aligned_condition_row["target"]:
                ordered_hidden_states.append(values)

    ordered_hidden_states = np.stack(ordered_hidden_states, axis=0)
    print(ordered_hidden_states.shape)
    rdm_by_layer = []
    for layer_num in range(ordered_hidden_states.shape[1]):
        layer_hidden_states = ordered_hidden_states[:, layer_num, :]
        rdm_layer_model = mne_rsa.compute_rdm(layer_hidden_states, metric="correlation")
        rdm_by_layer.append(rdm_layer_model)
    model_rdm_dict[model] = rdm_by_layer
```

$$
H_0: RSA(t) = 0 \forall
$$

В качестве альтернативной гипотезы $H_1$ выбираем:

$$
H_1: \exists t \in [t_{min}, t_{max}]: RSA(t) ~= 0
$$

Для проверки статистической значимости между матрицами субъектов и

```python
for s, el in enumerate(evokeds_list):
    evoked_list = [el[c] for c in conditions]
    labels = conditions

    for l, rdm_layer in enumerate(model_rdm_list):

        rsa_evoked = mne_rsa.rsa_evokeds(
            evokeds=evoked_list,
            rdm_model=rdm_layer,
            labels_evokeds=labels,
            labels_rdm_model=labels,
            spatial_radius=None,
            temporal_radius=0.010,
            evoked_rdm_metric="correlation",
            rsa_metric="spearman",
            picks=["Fz", "FCz", "Pz"],
            n_jobs=1,
            verbose=True,
            tmin=0.0,
            tmax=0.8
        )

        path_to_save_rsa_evoked = f"./rsa_evoked/sbj-{s}-layer-{l}-ave.fif"

        mne.write_evokeds(
            path_to_save_rsa_evoked,
            rsa_evoked,
            overwrite=True
        )
```

Кластерно пермутационный тест

```python
from statsmodels.stats.multitest import fdrcorrection

evoked_path = "./rsa_evoked"

subjects_by_layers_dict: dict[int, list] = {}

sbj_layer_files = os.listdir(evoked_path)
sbj_layer_files = list(filter(lambda x: "layer" in x, sbj_layer_files))

for sbj_layer_file in sbj_layer_files:
    layer_num = int(sbj_layer_file.split("layer-")[1].split("-ave")[0])
    if layer_num not in subjects_by_layers_dict:
        subjects_by_layers_dict[layer_num] = []
    subjects_by_layers_dict[layer_num].append(sbj_layer_file)

sorted_layers_nums = sorted(list(subjects_by_layers_dict.keys()))
pvals_layers = []


for layer_num in sorted_layers_nums:

    subjects = subjects_by_layers_dict[layer_num]
    subjects = sorted([f.split("-layer")[0] for f in subjects], key=lambda x: int(x.split("sbj-")[1]))

    rsa_evokeds = []

    for sub in subjects:
        evk = mne.read_evokeds(
            evoked_path + "/" + sub + "-layer-" + str(layer_num) + "-ave.fif",
            condition=0,
            verbose=False
        )
        rsa_evokeds.append(evk)

    from mne.stats import permutation_cluster_1samp_test
    X = np.array([evk.data.squeeze() for evk in rsa_evokeds])

    times = rsa_evokeds[0].times

    T_obs, clusters, cluster_p_values, H0 = permutation_cluster_1samp_test(
        X,
        n_permutations=5000,
        threshold=None,     # автоматический t-threshold
        tail=0,             # двусторонний тест
        adjacency=None,     # только время → adjacency не нужна
        out_type="mask",
        verbose=False
    )

    pvals_layers.append(cluster_p_values.min() if len(cluster_p_values) > 0 else 1.0)
    print(cluster_p_values)
    alpha = 0.05

    significant_clusters = [
        clusters[i]
        for i, p in enumerate(cluster_p_values)
        if p < alpha
    ]


    print("significant clusters" + str(significant_clusters))

    mean_rsa = X.mean(axis=0)

    import matplotlib.pyplot as plt

    plt.figure(figsize=(10, 4))
    plt.plot(times, mean_rsa, label=f"Mean RSA layer {layer_num}")
    plt.axhline(0, color="black", linestyle="--", linewidth=1)

    for cluster in significant_clusters:
        fb = np.zeros(times.shape[0], dtype=bool)
        fb[cluster] = True
        plt.fill_between(
            times,
            mean_rsa,
            where=fb,
            color="red",
            alpha=0.3
        )

    plt.xlabel("Time (s)")
    plt.ylabel("RSA (Spearman ρ)")
    plt.title("Group-level RSA (cluster-corrected)")
    plt.legend()
    plt.tight_layout()
    plt.show()

```
