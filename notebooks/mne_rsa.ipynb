```python
import os

import mne
# import mne_rsa
```

```python
epochs_list = []

for file_name in os.listdir("../hf_datasets/eeg"):
    if file_name.endswith(".fif"):
        epochs = mne.read_epochs(os.path.join("../hf_datasets/eeg", file_name))
        epochs = epochs.apply_baseline((-0.4, 0.0))
        epochs._data *= -1
        epochs_list.append(epochs)
```

```python
import numpy as np
import matplotlib.pyplot as plt

s_1_x_stimuli = [
    # "Stimulus/S  1_1",
    # "Stimulus/S  1_2",
    "Stimulus/S  1_3",
]

s_2_x_stimuli = [
    #"Stimulus/S  2_1",
    # "Stimulus/S  2_2",
    "Stimulus/S  2_3",
]

s_3_x_stimuli = [
    #"Stimulus/S  3_1",
     # "Stimulus/S  3_2",
    "Stimulus/S  3_3",
]

s_4_x_stimuli = [
    #"Stimulus/S  4_1",
 # "Stimulus/S  4_2",
    "Stimulus/S  4_3",
]

s_1_x_erps = [s[s_1_x_stimuli].average() for s in epochs_list]
s_2_x_erps = [s[s_2_x_stimuli].average() for s in epochs_list]
s_3_x_erps = [s[s_3_x_stimuli].average() for s in epochs_list]
s_4_x_erps = [s[s_4_x_stimuli].average() for s in epochs_list]

s_1_x_erps_ga: mne.EvokedArray = mne.grand_average(s_1_x_erps)
s_2_x_erps_ga: mne.EvokedArray = mne.grand_average(s_2_x_erps)
s_3_x_erps_ga: mne.EvokedArray = mne.grand_average(s_3_x_erps)
s_4_x_erps_ga: mne.EvokedArray = mne.grand_average(s_4_x_erps)



evokeds = [s_1_x_erps_ga, s_2_x_erps_ga, s_3_x_erps_ga, s_4_x_erps_ga]
labels  = ["Cong", "Semantic", "Gramatic", "Both"]

# channels = ['Fz', 'FCz', 'Pz', 'Oz']
channels = ['Fz']

fig, ax = plt.subplots()

for evk, label in zip(evokeds, labels):
    data = evk.copy().pick_channels(channels).data.mean(axis=0)  # среднее по выбранным каналам
    ax.plot(evk.times * 1000, data, label=label)

ax.set_title("Grand averages for selected channels")
ax.set_xlabel("Time (ms)")
ax.set_ylabel("Amplitude (µV)")
ax.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# параметры окна
win_size = 0.1   # 100 ms
step = 0.1      # 100 ms шаг
channel = 'Fz'

colors = ['blue', 'green', 'red', 'purple']  # цвета для условий

# извлечём частоту дискретизации
sfreq = evokeds[0].info['sfreq']
win_samples = int(win_size * sfreq)
step_samples = int(step * sfreq)

# получим данные по условиям
evoked_data = []
for evk in evokeds:
    data = evk.copy().pick_channels([channel]).data[0]
    evoked_data.append(data)

times = evokeds[0].times

# количество окон
num_windows = (len(times) - win_samples) // step_samples

for w_idx in range(num_windows):
    start = w_idx * step_samples
    end = start + win_samples

    t0 = times[start] * 1000
    t1 = times[end] * 1000

    # создаём гистограмму для данного окна
    plt.figure(figsize=(8, 5))
    plt.title(f"Window {t0:.0f}–{t1:.0f} ms (channel {channel})")

    for data, label, color in zip(evoked_data, labels, colors):
        w = data[start:end]
        plt.hist(w, bins=20, alpha=0.5, label=label, color=color)

    plt.xlabel("Amplitude (µV)")
    plt.ylabel("Count")
    plt.legend()
    plt.tight_layout()
    plt.show()
```
